# Setup project environment for formula1 project
- Create external location databricks_course_ext_dl1_formula1
- Create catalog formula1
- Create schema landing,bronze,silver and gold
- Create Volume Files in the landing schema


```sql
%fs ls 'abfss://formula1@databrickscourseextdela.dfs.core.windows.net/'
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_course_ext_dl1_formula1
URL 'abfss://formula1@databrickscourseextdela.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `databricks-course-sc`)
COMMENT 'External location for formula1 container';
```

```sql
SHOW CATALOGS;
```

```sql
CREATE CATALOG  IF NOT EXISTS formula1_catalog
    MANAGED LOCATION 'abfss://formula1@databrickscourseextdela.dfs.core.windows.net/'
   COMMENT 'Main catalog for formula1 project' ;
```

```sql
CREATE SCHEMA   IF NOT EXISTS  formula1_catalog.landing;
CREATE SCHEMA   IF NOT EXISTS  formula1_catalog.bronze
     MANAGED LOCATION 'abfss://formula1@databrickscourseextdela.dfs.core.windows.net/bronze';
CREATE SCHEMA   IF NOT EXISTS  formula1_catalog.silver
     MANAGED LOCATION 'abfss://formula1@databrickscourseextdela.dfs.core.windows.net/silver';
CREATE SCHEMA   IF NOT EXISTS  formula1_catalog.gold
     MANAGED LOCATION 'abfss://formula1@databrickscourseextdela.dfs.core.windows.net/gold';
```

```sql
CREATE EXTERNAL VOLUME formula1_catalog.landing.files
LOCATION 'abfss://formula1@databrickscourseextdela.dfs.core.windows.net/landing';
```

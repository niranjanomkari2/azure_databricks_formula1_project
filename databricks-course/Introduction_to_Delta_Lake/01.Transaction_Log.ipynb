##Understanding Delta Lake Transaction Log
    Understand cloud storage structure for a delta table

## Create catalog and schema for the Demo
- Catalog : demo
- Schema : delta_lake

```sql
CREATE CATALOG IF NOT EXISTS demo
MANAGED LOCATION 'abfss://demo@databrickscourseextdela.dfs.core.windows.net/'
```

```sql
CREATE SCHEMA IF NOT EXISTS demo.delta_lake
MANAGED LOCATION 'abfss://demo@databrickscourseextdela.dfs.core.windows.net/delta_lake'
```

### 1.Create a delta lake table

```sql
CREATE TABLE IF NOT EXISTS demo.delta_lake.companies
(
    company_name STRING,
    founded_date DATE,
    country STRING
)
USING DELTA;
```

```sql
DESC EXTENDED demo.delta_lake.companies 
```

```sql
insert into demo.delta_lake.companies
values ('Microsoft','1975-04-04','USA'),
('Google','1998-09-04','USA'),
('Amazon','1994-07-04','USA');
```

```sql
SELECT * FROM demo.delta_lake.companies
```

#H2 Build Races Dimension
- Read Silver races table
- Read silver circuits table
- Join data from races and circuits using circuit_id
- Select the required columns
    - races.season
    -  races.round
    -  races.race_name
    -  races.race_date
    -  circuits.circuit_name
    -  circuits.locality
    -  circuits.country
- Write the transformed data to gold dim_races table

```python
%run ../00-common/01.environment_config
```

```python
target_table = f'{catalog_name}.{gold_schema}.dim_races'
display(target_table)
```

## Read Source tables
- races
- circuits

```python
races_df = spark.table(f"{catalog_name}.{silver_schema}.races")
circuits_df = spark.table(f"{catalog_name}.{silver_schema}.circuits")
```

##Step2 - Join races with circuits using circuit_id
    races.season
    races.round
    races.race_name
    races.race_date
    circuits.circuit_name
    circuits.locality
    circuits.country_name

```python
dim_races_df =  (
    races_df
    .join(
        circuits_df,
        races_df.circuit_id == circuits_df.circuit_id,
        "inner"
        )
    .select(
        races_df.season,
        races_df.round,
        races_df.race_name,
        races_df.race_date,
        circuits_df.circuit_name,
        circuits_df.locality,
        circuits_df.country_name
           )
                )
```

```python
display(dim_races_df)
```

```python
(
    dim_races_df
.write \
.format('delta') \
.mode("overwrite")
.saveAsTable(target_table)
)
```
